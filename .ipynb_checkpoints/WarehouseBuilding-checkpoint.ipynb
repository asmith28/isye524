## SOS Example -- Building a warehouse

We are modeling a facility location problem and must decide what size of warehouse to build. The choices of sizes and their associated costs are as follows:


```julia
size = [10 20 40 60 80]
cost = [100 180 320 450 600];
```

```julia
using JuMP, Gurobi

m = Model(solver=GurobiSolver(OutputFlag=0))

@variable(m, x[1:5] >= 0)
@variable(m, wsize >= 0)
addSOS1(m, x)

@objective(m, Min, sum(x[i]*cost[i] for i in 1:5))

@constraint(m, wsize == sum(x[i]*size[i] for i in 1:5))
@constraint(m, sum(x) == 1)

solve(m)
println(getvalue(x))
println("Chosen value: (", x[1], ", ", getvalue(wsize), ")")
```
