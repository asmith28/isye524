## Getting Started with Julia -- Solution

```julia
using JuMP, Clp, ECOS, SCS

m = Model()

@variable(m, 1 <= x1 <= 4)
@variable(m, 0 <= x2 <= 2)
@variable(m, 0 <= x3 <= 2)

@objective(m, Max, -4x1 + 2x2 + 12x3)

@constraint(m, x1 >= x2 + x3)

setsolver(m, ClpSolver())

@time(solve(m))

println("Objective: ", getobjectivevalue(m))
println("x1: ", getvalue(x1))
println("x2: ", getvalue(x2))
println("x3: ", getvalue(x3))

setsolver(m, ECOSSolver(verbose=0))

@time(solve(m))

println("Objective: ", getobjectivevalue(m))
println("x1: ", getvalue(x1))
println("x2: ", getvalue(x2))
println("x3: ", getvalue(x3))

setsolver(m, SCSSolver(verbose=0))

@time(solve(m))

println("Objective: ", getobjectivevalue(m))
println("x1: ", getvalue(x1))
println("x2: ", getvalue(x2))
println("x3: ", getvalue(x3))
```

The solver times are comparable, but Clp typically performs slightly faster. This is because Clp is a linear programming solver, whereas SCS and ECOS are conic solvers and do a lot of extra work to convert the model into a specific form. (The problem is small enough that the differences might not be noticable, so your interpretation of the outcome could vary a bit.) Clp gets the true answer (2,0,2) and is more accurate than the other solvers, for the same reason as above.

## Terraforming Mars -- Solution

### (a) Math formulation

Decision variables:

$x_A$ = square kilometers allocated to algae type A

$x_B$ = square kilometers allocated to algae type B

Objective: 

$\max \ 300x_A + 200x_B$

Constraints:

Available space: $x_A + x_B \leq 100$

Budget: $2x_A + x_B \leq 180$

Nonnegativity: $x_A, x_B \geq 0$


### Julia code

```julia
using JuMP, Clp

m = Model(solver=ClpSolver())

@variable(m, xA >= 0)
@variable(m, xB >= 0)

@objective(m, Max, 300xA + 200xB)

@constraint(m, xA + xB <= 100)
@constraint(m, 2xA + xB <= 180)

solve(m)

println("Oxygen production: ", getobjectivevalue(m))
println("Square kilometers of algae A: ", getvalue(xA))
println("Square kilometers of algae A: ", getvalue(xB))
```

### (b) Modular version

```julia
# Data
types = [:A, :B]

oxygen = Dict(zip(types,[300, 200]))

cost = Dict(zip(types,[2, 1]))

space_avail = 100
budget_avail = 180

# Model
m = Model(solver=ClpSolver())

@variable(m, x[types] >= 0)

@objective(m, Max, sum(oxygen[t]*x[t] for t in types))

@constraint(m, sum(x[t] for t in types) <= space_avail)
@constraint(m, sum(cost[t]*x[t] for t in types) <= budget_avail)

solve(m)

println("Oxygen production: ", getobjectivevalue(m))
println("Square kilometers of each type: ", getvalue(x))
```
