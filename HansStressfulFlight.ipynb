```julia
using JuMP, Clp

m = Model(solver=ClpSolver())

@variable(m, x >= 0) # speed Han flies
@variable(m, t) # epigraph trick variable

@objective(m, Min, t)

@constraint(m, t >= -2*x + 20) # Leia's function
@constraint(m, t >= -(1/10)*x + 10) # Luke's function
@constraint(m, t >= x + 3) # C3P0's function
@constraint(m, x >= 6) 
@constraint(m, x <= 12) 

solve(m)

println("Han should fly at ", getvalue(x), "parsecs/min")
println("Minimax stress level: ",getvalue(t))
```
