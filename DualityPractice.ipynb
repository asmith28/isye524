## Duality practice from July 6

Consider the following linear program:

$\begin{align*}
\max_x \ & -4x_1 - x_2 \\ 
\text{s.t.} \ & 4x_1 + 3x_2 \geq 6\\
& x_1 + 2x_2 \leq 3\\
& 3x_1 + x_2 = 3\\
& x_1, x_2 \geq 0.
\end{align*}$

* Write down the dual of this linear program. Look at lecture 7, slide 15 for the list of dual rules. 
* Suppose we had the option of paying \$2/unit to increase the right-hand side of the second constraint. Would this be a good deal for us?

### Dual, following dual rules table

* Need one dual variable for each primal constraint (3 dual variables). Call them $\lambda_1, \lambda_2, \lambda_3$. 
* Dual is minimization
* Right-hand side of primal constraints are dual costs
* Right-hand side of dual constraints are primal costs
* Constraint matrix of dual is transpose of constraint matrix of primal
* Nonnegative primal variables $(x_1,x_2)$ correspond to dual constraints that are $\geq$ (in a minimization problem)
* $\lambda_1$ must be $\leq 0$ since the corresponding primal constraint is $\geq$ and it is a maximization problem
* $\lambda_2$ must be $\geq 0$ since the corresponding primal constraint is $\leq$ and it is a maximization problem
* $\lambda_3$ must be free since the corresponding primal constraint is $=$

$\begin{align*}
\min_{\lambda}\ & 6\lambda_1 + 3\lambda_2 + 3\lambda_3\\
\text{s.t.} \ & 4\lambda_1 + \lambda_2 + 3\lambda_3 \geq -4\\
& 3\lambda_1 + 2\lambda_2 + \lambda_3 \geq -1\\
& \lambda_1 \leq 0\\
& \lambda_2 \geq 0 \\
& \lambda_3 \text{ free}
\end{align*}$

### Getting the shadow price of the 2nd primal constraint

```julia
# We can solve the dual problem directly to get the value of λ2 
using JuMP, Clp

m = Model(solver=ClpSolver())

@variable(m, λ1 <= 0)
@variable(m, λ2 >= 0)
@variable(m, λ3)

@objective(m, Min, 6λ1 + 3λ2 + 3λ3)

@constraint(m, 4λ1 + λ2 + 3λ3 >= -4)
@constraint(m, 3λ1 + 2λ2 + λ3 >= -1)

solve(m)

println("Shadow price of second primal constraint: ", getvalue(λ2)) # shadow prices are dual variable values
```

Paying \$2 per unit for an increase in the right-hand side of the second primal constraint is NOT a good deal for us, since we would only be willing to pay \$0.20 per unit (each unit we buy increases our profit by \$0.20, so paying \$2 per unit would be a \$1.80 loss for every unit we purchase).

```julia
# We can also solve the primal model and call "getdual" to get the same outcome
using JuMP, Clp

m = Model(solver=ClpSolver())

@variable(m, x1 >= 0)
@variable(m, x2 >= 0)

@objective(m, Max, -4x1 - x2)

@constraint(m, c1, 4x1 + 3x2 >= 6)
@constraint(m, c2, x1 + 2x2 <= 3)
@constraint(m, c3, 3x1 + x2 == 3)

solve(m)

println("Shadow price of second primal constraint: ", getdual(c2)) # getdual(constraint) gets shadow price of that constraint
```
