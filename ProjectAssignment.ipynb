## Assigning students to projects

A teacher needs to assign each of 5 projects to 5 different students. Each student has given a preference for each project by giving each project a rating from 0 to 10 (0 indicates strong dislike and 10 indicates strong preference). The teacher wishes to assign projects in such a way as to maximize the overall satisfaction. The preferences of the students are as follows:

| Project| Akhil | Brianna | Carlos | Dinah | Edward |
|-------:|------:|--------:|-------:|------:|-------:|
|      1 | 2     |   9     |  1     | 9     | 6      |  
|      2 | 9     |   0     |  5     | 5     | 8      |  
|      3 | 7     |   1     |  10    | 2     | 6      |
|      4 | 0     |   7     |  8     | 1     | 7      |
|      5 | 4     |   0     |  4     | 0     | 4      |


```julia
using JuMP, Clp, NamedArrays

projects = [ :1, :2, :3, :4, :5 ]
students = [ :Akhil, :Brianna, :Carlos, :Dinah, :Edward ]

raw = [ 2 9 1 9 6
        9 0 5 5 8
        7 1 10 2 6
        0 7 8 1 7
        4 0 4 0 4]

preferences = NamedArray( raw, (projects,students), ("proj","stud"))

m = Model(solver=ClpSolver())

@variable(m, x[projects,students] >= 0)

# each student gets exactly 1 project
@constraint(m, a[j in students], sum(x[i,j] for i in projects) == 1 )

# each project gets assigned to one student
@constraint(m, b[i in projects], sum(x[i,j] for j in students) == 1 )

@objective(m, Max, sum( x[i,j]*preferences[i,j] for i in projects, j in students ) )

solve(m)

assignment = NamedArray( [ (getvalue(x[i,j])) for i in projects, j in students ], (projects, students), ("proj","stud"))

```

Let this be a lesson about the dangers of using neutral numbers!
