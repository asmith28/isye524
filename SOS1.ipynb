## SOS1 Example -- Multiple Choice

Suppose we have the following possible (x,y) pairs:
![sos1](https://raw.githubusercontent.com/asmith28/isye524/master/sos1.png)

```julia
x = [1.5 2.1 4.5 6.8 7.1 9.4]
a = [3 2 4 1 3 2]

using JuMP, Gurobi

m = Model(solver=GurobiSolver(OutputFlag=0))

@variable(m, λ[1:6] >= 0)
addSOS1(m, λ)

@variable(m, y >= 0)
@constraint(m, y == sum(a[i]*λ[i] for i in 1:6))
@constraint(m, sum(λ) == 1)

solve(m)
println(getvalue(λ))
println("Chosen value: (", x[6], ", ", getvalue(y), ")")
```
