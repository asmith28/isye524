## Building a house (Harvard Business Review, 1963)


Several tasks must be completed in order to build our house. Each task has a duration, tasks may be worked on simultaneously, but there is also a precedence relation. Some tasks can only be started once other tasks have been completed. The following graph shows each task, its duration, and the tasks that must be completed before it starts. How fast can the house be built?
